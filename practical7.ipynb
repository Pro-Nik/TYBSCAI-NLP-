# SMS Spam Text Classification

# Step 1: Import Libraries

```python
# Import pandas for working with CSV data
import pandas as pd

# Convert text into numerical form
from sklearn.feature_extraction.text import TfidfVectorizer

# Split data into training and testing data
from sklearn.model_selection import train_test_split

# Naive Bayes classification algorithm
from sklearn.naive_bayes import MultinomialNB

# Evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

# Step 2: Load Your Dataset

```python
# Load the spam.csv file

df = pd.read_csv(
    "spam.csv",
    encoding="latin-1"
)

# Display first 5 rows

print(df.head())
```

# Step 3: Check All Column Names

```python
print(df.columns)
```

# Step 4: Keep Only Required Columns

```python
# Keep only label and message columns

df = df[["v1", "v2"]]
```

```python
# Rename columns

df.columns = [
    "label",
    "message"
]

# Display first 5 rows

print(df.head())
```

# Step 5: Check Dataset Information

```python
# Check dataset shape

print("Dataset Shape:")
print(df.shape)


# Check missing values

print("\nMissing Values:")
print(df.isnull().sum())


# Check Spam and Ham counts

print("\nClass Distribution:")
print(df["label"].value_counts())
```

# Step 6: Separate Input and Output

```python
# X contains SMS messages

X = df["message"]


# y contains labels

y = df["label"]
```

# Step 7: Train-Test Split

```python
# Divide data into training and testing data

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    # 20% data for testing
    test_size=0.2,

    # Same result every time
    random_state=42,

    # Maintain Spam/Ham ratio
    stratify=y
)
```

# Step 8: Convert Text into Numbers Using TF-IDF

```python
# Create TF-IDF vectorizer

vectorizer = TfidfVectorizer()
```

```python
# Learn vocabulary from training data
# and convert messages into numerical vectors

X_train_tfidf = vectorizer.fit_transform(X_train)
```

```python
# Convert test messages using the same vocabulary

X_test_tfidf = vectorizer.transform(X_test)
```

# Step 9: Create Naive Bayes Model

```python
# Create Naive Bayes model

model = MultinomialNB()
```

# Step 10: Train the Model

```python
# Train the model

model.fit(
    X_train_tfidf,
    y_train
)
```

# Step 11: Make Predictions

```python
# Predict test data

y_pred = model.predict(
    X_test_tfidf
)
```

# Step 12: Compare Actual and Predicted Values

```python
# Create a DataFrame to compare results

results = pd.DataFrame({

    "Actual": y_test.values,

    "Predicted": y_pred
})


print(results.head(10))
```

# Step 13: Calculate Accuracy

```python
# Calculate accuracy

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:", accuracy)
```

```python
print("Accuracy Percentage:", accuracy * 100)
```

# Step 14: Confusion Matrix

```python
# Create confusion matrix

cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)
```

```python
# Enter a new message

new_message = [

    "Congratulations you have won a free prize"
]
```
