# Pre-Processing

```python
text = """Natural Language Processing (NLP) is one of the most exciting fields of
Artificial Intelligence. It enables computers to understand, interpret,
and generate human language. Today, NLP powers many applications such as
chatbots, virtual assistants, machine translation systems, sentiment
analysis tools, and search engines.

Students who study NLP learn how computers process text data collected
from websites, books, research papers, social media platforms, and
customer reviews. However, raw text is often noisy and contains
punctuation marks, special characters, uppercase letters, and irrelevant
words. Therefore, before building machine learning models, the text must
be cleaned and preprocessed carefully.

The preprocessing pipeline generally includes tokenization,
lowercasing, stopword removal, stemming, and lemmatization. These
techniques help reduce vocabulary size, remove redundancy, and improve
the efficiency of NLP algorithms. Modern transformer-based models such
as BERT and GPT use advanced tokenization techniques, but understanding
the classical preprocessing pipeline remains essential for every NLP
practitioner."""
```

# Find no. of characters and words

```python
print("Number of Characters:", len(text))
print("Number of Words:", len(text.split()))
```

# Sentence Tokenization

```python
import nltk

nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(text)

for i, sentence in enumerate(sentences):
    print(f"Sentence {i+1}: {sentence}")
```

# Word Tokenization

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)

print(tokens[:10])
```

```python
print("Total Tokens:", len(tokens))
```

# Lowercasing

```python
lower_tokens = [word.lower() for word in tokens]

print(lower_tokens[:20])
```

# Voc size before

```python
vocab_before = set(tokens)
print(len(vocab_before))
```

# Voc size after

```python
vocab_after = set(lower_tokens)
print(len(vocab_after))
```

# Remove Punctuation

```python
import string

tokens_no_punct = [
    word for word in lower_tokens
    if word not in string.punctuation
]
```

```python
print(tokens_no_punct)
```

# Stopwords

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens_no_punct
    if word not in stop_words
]
```

```python
filtered_tokens
```

```python
print(len(set(tokens_no_punct)))
print(len(set(filtered_tokens)))
```

# Frequency distribution

```python
from collections import Counter

freq = Counter(filtered_tokens)

print(freq.most_common(5))
```

# Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stems = [
    stemmer.stem(word)
    for word in filtered_tokens
]
```

```python
stems
```

# Lemmatization

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmas = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]
```

```python
lemmas
```

```python
len(lemmas)
```

# Save the final corpus for the text-representation

```python
final_corpus = " ".join(lemmas)

with open("final_corpus.txt", "w", encoding="utf-8") as file:
    file.write(final_corpus)

print("Corpus saved successfully!")
```
