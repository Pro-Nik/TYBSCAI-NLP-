```python
from gensim.models import Word2Vec
```

```python
sentences = [
    ["machine","learning","is","powerful"],
    ["machine","learning","uses","data"],
    ["deep","learning","uses","neural","networks"],
    ["artificial","intelligence","uses","machine","learning"],
    ["python","is","used","for","machine","learning"],
    ["deep","learning","is","part","of","artificial","intelligence"]
]

print(sentences)
```

```python
model = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=4,
    sg=0
)
```

```python
print(model.wv.index_to_key)
```

```python
vector = model.wv["machine"]

print(vector)
```

```python
vector = model.wv["machine"]

print(vector)
```

```python
model.wv.most_similar("machine")
```

```python
similarity = model.wv.similarity("machine","learning")

print(similarity)
```

```python
model.wv.doesnt_match(["machine","learning","deep","apple"])
```

```python
model.save("word2vec_model.model")
```

```python
from gensim.models import Word2Vec

loaded_model = Word2Vec.load("word2vec_model.model")
```

```python
for word in loaded_model.wv.index_to_key:
    print(word)
    print(loaded_model.wv[word])
    print("-"*50)
```
